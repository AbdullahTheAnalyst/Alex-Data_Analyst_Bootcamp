```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\otabek.khamidov\Desktop\Desktop\Data Analytics\Alex-Data_Analyst_Bootcamp\Pandas\Ice Cream Ratings.csv")
df = df.set_index("Date")
df
```

```python
# kind orqali chart turini beramiz
df.plot(kind = "line")
plt.show()
```

```python
# subplots orqali chart har bir qismga ajratib beradi
df.plot(kind = "line", subplots = True)
plt.show()
```

```python
# title hamda label.larni belgilashimiz uchun
df.plot(kind = "line", title = "Ice Cream Ratings", xlabel = "Daily Ratings", ylabel = "Scores")
plt.show()
```

```python
# bar plot:
df.plot(kind = "bar")
plt.show()
```

```python
# bar plot only one data:
df["Texture Rating"].plot(kind = "bar", title = "Only Texture Rating", xlabel = "Daily Ratings for TR", ylabel = "Scores")
plt.show()
```

```python
# horizontal chizish uchun
df.plot.barh(stacked  = True)
# or df.plot(kind = "barh", stacked  = True)
plt.show()
```

```python
# scatterplot (bog'liqligi) chizish
df.plot(kind = "scatter", x = "Flavor Rating", y = "Overall Rating", s = 100, c = "Red")
```

```python
df.plot(kind = "hist", bins = 100)
plt.show()
```

```python
df.boxplot()
plt.show()
```

```python
df.plot(kind = "area", figsize = (15, 5))
plt.show()
```

```python
df.plot(kind = "pie", y = "Flavor Rating", figsize = (15, 15))
plt.show()
```
