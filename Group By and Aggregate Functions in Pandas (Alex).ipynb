```python
import pandas as pd

# Ma'lumotlar
data = {
    'yosh': [23, 45, 12, 36, 52],
    'daromad': [5000, 7000, 2000, 6500, 8000]
}
df = pd.DataFrame(data)

# Aggregation
natija = df.aggregate({
    'yosh': ['min', 'max', 'mean'],
    'daromad': ['min', 'max', 'mean']
})

print(natija)
```

```python
df.groupby("Base Flavor").agg({
    "Flavor Rating" : ["min", "max", "sum", "mean"]
})
```

```python
df.groupby("Base Flavor").agg({
    "Flavor Rating" : ["min", "max", "sum", "mean"],
    "Texture Rating" : ["min", "max", "sum", "mean"]
})
```
